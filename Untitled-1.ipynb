```python
# Extract floor level from 'door' column
def extract_floor(s):
    """Extracts the floor level from a 'door' entry, if available."""
    if isinstance(s, str):
        match = re.match(r'^(\d{1,2})º', s)
        return int(match.group(1)) if match else None
    return None

data['floor'] = data['door'].apply(extract_floor)
```

```python
# Fix orientation
data["orientation"] = data["orientation"].replace("soxth", "south")
```

```python
# NEW: One-hot encode orientation and neighborhood
data = pd.get_dummies(data, columns=["orientation", "neighborhood"], dummy_na=True)
```

```python
# Convert categorical columns to numerical codes
def preprocess_data(data):
    """Converts categorical features to numerical codes."""
    # categorical_columns = ["has_pool", "is_furnished", "orientation", "has_ac", "accepts_pets", "neighborhood"]
    categorical_columns = ["has_pool", "is_furnished", "has_ac", "accepts_pets"]
    for column in categorical_columns:
        if column in data.columns:
            data[column] = data[column].astype('category').cat.codes
        else:
            raise ValueError(f"Column '{column}' is missing in the dataset.")
    return data

data = preprocess_data(data).drop(columns=['door'])
```

```python
# Replace outliers with median for 'num_rooms'
def replace_outliers_with_median(data, column, threshold):
    """Replaces values greater than a threshold with the median value."""
    median_value = data[column].median()
    data.loc[data[column] > threshold, column] = median_value
    return data

# Replace negative values with median for specified column
def replace_negative_with_median(data, column):
    """Replaces negative values in a column with the median value of non-negative entries."""
    median_value = data.loc[data[column] >= 0, column].median()
    data.loc[data[column] < 0, column] = median_value
    return data

# Apply imputations and corrections to training data
train_data = replace_outliers_with_median(train_data, 'num_rooms', 10)
train_data = replace_negative_with_median(train_data, 'square_meters')
```

# Feature analysis

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, PolynomialFeatures
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.feature_selection import mutual_info_regression
from sklearn.inspection import permutation_importance
import matplotlib.pyplot as plt
import seaborn as sns
from tqdm import tqdm
from scipy import stats

class PerformanceAnalysis:
    def __init__(self, random_state=42):
        self.random_state = random_state
        
    def analyze_error_distribution(self, X, y):
        """
        Analyze the distribution of errors across different models and feature combinations
        """
        # Split data
        X_train, X_test, y_train, y_test = train_test_split(
            X, y, test_size=0.2, random_state=self.random_state
        )
        
        # Scale features
        scaler = StandardScaler()
        X_train_scaled = scaler.fit_transform(X_train)
        X_test_scaled = scaler.transform(X_test)
        
        # Basic model for baseline
        rf = RandomForestRegressor(n_estimators=100, random_state=self.random_state)
        rf.fit(X_train_scaled, y_train)
        y_pred = rf.predict(X_test_scaled)
        
        # Analyze residuals
        residuals = y_test - y_pred
        
        # Plot residual analysis
        plt.figure(figsize=(15, 5))
        
        # Residual distribution
        plt.subplot(131)
        sns.histplot(residuals, kde=True)
        plt.title('Residual Distribution')
        plt.xlabel('Residual Value')
        
        # Q-Q plot
        plt.subplot(132)
        from scipy import stats
        stats.probplot(residuals, dist="norm", plot=plt)
        plt.title('Q-Q Plot of Residuals')
        
        # Residuals vs Predicted
        plt.subplot(133)
        plt.scatter(y_pred, residuals, alpha=0.5)
        plt.axhline(y=0, color='r', linestyle='--')
        plt.title('Residuals vs Predicted')
        plt.xlabel('Predicted Values')
        plt.ylabel('Residuals')
        
        plt.tight_layout()
        plt.show()
        
        return residuals
    
    def feature_importance_analysis(self, X, y):
        """
        Analyze feature importance using multiple methods
        """
        # Convert to numpy arrays
        X_values = X.values if hasattr(X, 'values') else X
        feature_names = X.columns if hasattr(X, 'columns') else [f'Feature_{i}' for i in range(X.shape[1])]
        
        # Calculate mutual information
        mi_scores = mutual_info_regression(X_values, y)
        
        # Random Forest importance
        rf = RandomForestRegressor(n_estimators=100, random_state=self.random_state)
        rf.fit(X_values, y)
        rf_importance = rf.feature_importances_
        
        # Permutation importance
        perm_importance = permutation_importance(rf, X_values, y, 
                                               n_repeats=10, 
                                               random_state=self.random_state)
        
        # Create importance DataFrame
        importance_df = pd.DataFrame({
            'Feature': feature_names,
            'Mutual Information': mi_scores,
            'Random Forest': rf_importance,
            'Permutation': perm_importance.importances_mean
        })
        
        # Plot importance scores
        plt.figure(figsize=(12, 6))
        importance_df_melted = pd.melt(importance_df, 
                                     id_vars=['Feature'], 
                                     var_name='Method',
                                     value_name='Importance')
        
        sns.barplot(data=importance_df_melted, 
                   x='Feature', 
                   y='Importance', 
                   hue='Method')
        plt.xticks(rotation=45)
        plt.title('Feature Importance by Different Methods')
        plt.tight_layout()
        plt.show()
        
        return importance_df
    
    def analyze_feature_interactions(self, X, y):
        """
        Analyze potential feature interactions
        """
        # Create polynomial features
        poly = PolynomialFeatures(degree=2, include_bias=False)
        X_poly = poly.fit_transform(X)
        
        # Get feature names
        if hasattr(X, 'columns'):
            feature_names = list(X.columns)
            poly_features = poly.get_feature_names_out(feature_names)
        else:
            feature_names = [f'Feature_{i}' for i in range(X.shape[1])]
            poly_features = poly.get_feature_names_out(feature_names)
        
        # Fit model with interactions
        rf = RandomForestRegressor(n_estimators=100, random_state=self.random_state)
        rf.fit(X_poly, y)
        
        # Get importance scores for interaction terms
        importance = pd.DataFrame({
            'Feature': poly_features,
            'Importance': rf.feature_importances_
        })
        
        # Filter for interaction terms only
        interaction_terms = importance[importance['Feature'].str.contains(' ')]
        interaction_terms = interaction_terms.sort_values('Importance', 
                                                        ascending=False)
        
        # Plot top interactions
        plt.figure(figsize=(12, 6))
        sns.barplot(data=interaction_terms.head(10), 
                   x='Importance', 
                   y='Feature')
        plt.title('Top 10 Feature Interactions')
        plt.tight_layout()
        plt.show()
        
        return interaction_terms
    
    def analyze_noise_impact(self, X, y, n_iterations=100):
        """
        Analyze the impact of noise on model performance
        """
        mse_scores = []
        noise_levels = np.linspace(0, 1, 10)
        
        for noise_level in tqdm(noise_levels):
            iteration_scores = []
            for _ in range(n_iterations):
                # Add noise to target
                noise = np.random.normal(0, noise_level * y.std(), size=len(y))
                y_noisy = y + noise
                
                # Split and evaluate
                X_train, X_test, y_train, y_test = train_test_split(
                    X, y_noisy, test_size=0.2, random_state=None
                )
                
                # Fit and evaluate model
                rf = RandomForestRegressor(n_estimators=100, 
                                         random_state=self.random_state)
                rf.fit(X_train, y_train)
                y_pred = rf.predict(X_test)
                mse = mean_squared_error(y_test, y_pred)
                iteration_scores.append(mse)
            
            mse_scores.append(iteration_scores)
        
        # Plot noise impact
        plt.figure(figsize=(10, 6))
        plt.boxplot(mse_scores)
        plt.xlabel('Noise Level')
        plt.ylabel('MSE')
        plt.title('Impact of Noise on Model Performance')
        plt.xticks(range(1, len(noise_levels) + 1), 
                  [f'{n:.2f}' for n in noise_levels])
        plt.xticks(rotation=45)
        plt.tight_layout()
        plt.show()
        
        return pd.DataFrame({
            'Noise_Level': noise_levels,
            'Mean_MSE': [np.mean(scores) for scores in mse_scores],
            'Std_MSE': [np.std(scores) for scores in mse_scores]
        })
    
    def suggest_improvements(self, X, y):
        """
        Analyze data and suggest potential improvements
        """
        # Get feature importance
        importance_df = self.feature_importance_analysis(X, y)
        
        # Get interaction analysis
        interaction_terms = self.analyze_feature_interactions(X, y)
        
        # Analyze residuals
        residuals = self.analyze_error_distribution(X, y)
        
        # Make suggestions based on analysis
        suggestions = []
        
        # Check for weak features
        weak_features = importance_df[importance_df['Random Forest'] < 
                                    importance_df['Random Forest'].mean() * 0.1]
        if not weak_features.empty:
            suggestions.append(
                "Consider removing or transforming these weak features: " + 
                ", ".join(weak_features['Feature'].tolist())
            )
        
        # Check for strong interactions
        strong_interactions = interaction_terms[
            interaction_terms['Importance'] > interaction_terms['Importance'].mean()
        ]
        if not strong_interactions.empty:
            suggestions.append(
                "Consider explicitly creating these interaction features: " + 
                ", ".join(strong_interactions['Feature'].head().tolist())
            )
        
        # Check residual distribution
        if stats.normaltest(residuals)[1] < 0.05:
            suggestions.append(
                "Residuals are not normally distributed. Consider:"
                "\n- Transforming the target variable"
                "\n- Using more robust models"
                "\n- Adding non-linear feature transformations"
            )
        
        return suggestions
```

```python
# Usage example:

# Initialize analysis
analysis = PerformanceAnalysis()

# Analyze residuals and error distribution
residuals = analysis.analyze_error_distribution(X_train, y_train)

# Analyze feature importance
importance_df = analysis.feature_importance_analysis(X_train, y_train)

# Analyze feature interactions
interaction_terms = analysis.analyze_feature_interactions(X_train, y_train)

# Analyze impact of noise
noise_analysis = analysis.analyze_noise_impact(X_train, y_train)

# Get improvement suggestions
suggestions = analysis.suggest_improvements(X_train, y_train)
print("\nSuggested Improvements:")
for i, suggestion in enumerate(suggestions, 1):
    print(f"{i}. {suggestion}")
```

```python
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder, RobustScaler, FunctionTransformer
from sklearn.compose import ColumnTransformer
from sklearn.impute import KNNImputer, SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.ensemble import IsolationForest

from keras import layers, models, callbacks
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

class RobustNeuralRegression:
    def __init__(self, random_state=42):
        self.random_state = random_state
        self.numeric_features = ['num_rooms', 'num_baths', 'square_meters', 
                               'year_built', 'num_supermarkets', 'num_crimes']
        self.categorical_features = ['orientation', 'door', 'neighborhood']
        self.boolean_features = ['is_furnished', 'has_pool', 'has_ac', 'accepts_pets']
        self.preprocessor = None
        self.model = None
        self.history = None
        
    def detect_outliers(self, X_numeric):
        """Detect outliers in numeric features using Isolation Forest"""
        print("\nDetecting outliers in numeric features...")
        
        # Initial imputation for outlier detection
        initial_imputer = SimpleImputer(strategy='median')
        X_numeric_imputed = initial_imputer.fit_transform(X_numeric)
        
        # Detect outliers
        iso_forest = IsolationForest(contamination=0.1, random_state=self.random_state)
        outlier_labels = iso_forest.fit_predict(X_numeric_imputed)
        outlier_scores = iso_forest.score_samples(X_numeric_imputed)
        
        # Convert to reliability scores (higher means more reliable)
        reliability_scores = (outlier_scores - outlier_scores.min()) / \
                           (outlier_scores.max() - outlier_scores.min())
        
        print(f"Found {(outlier_labels == -1).sum()} outliers")
        return reliability_scores.reshape(-1, 1)
    
    def create_preprocessor(self):
        """Create preprocessing pipeline"""
        print("\nCreating preprocessing pipeline...")
        
        # Numeric features pipeline
        numeric_pipeline = Pipeline([
            ('imputer', SimpleImputer(strategy='median')),
            ('scaler', RobustScaler())
        ])
        
        # Categorical features pipeline
        categorical_pipeline = Pipeline([
            ('imputer', SimpleImputer(strategy='constant', fill_value='missing')),
            ('onehot', OneHotEncoder(drop='first', sparse_output=False, handle_unknown='ignore'))
        ])
        
        # Boolean features pipeline
        boolean_pipeline = Pipeline([
            ('imputer', SimpleImputer(strategy='most_frequent')),
            ('bool_to_float', FunctionTransformer(lambda x: x.astype(float)))
        ])
        
        self.preprocessor = ColumnTransformer(
            transformers=[
                ('num', numeric_pipeline, self.numeric_features),
                ('cat', categorical_pipeline, self.categorical_features),
                ('bool', boolean_pipeline, self.boolean_features)
            ],
            remainder='drop'
        )
        
    def prepare_data(self, X, y=None):
        """Prepare and preprocess data"""
        print("\nPreparing data...")
        
        # Convert to DataFrame if needed
        if not isinstance(X, pd.DataFrame):
            X = pd.DataFrame(X, columns=(
                self.numeric_features + 
                self.categorical_features + 
                self.boolean_features
            ))
        
        # Reset index
        X = X.reset_index(drop=True)
        
        # Convert boolean columns
        for col in self.boolean_features:
            X[col] = X[col].astype('boolean')
        
        # Convert numeric columns
        for col in self.numeric_features:
            X[col] = pd.to_numeric(X[col], errors='coerce')
        
        print("\nData types after conversion:")
        print(X.dtypes)
        
        # Create preprocessor if not exists
        if self.preprocessor is None:
            self.create_preprocessor()
        
        # Process all features first
        if y is not None:
            X_processed = self.preprocessor.fit_transform(X)
            print(f"Preprocessed feature shape: {X_processed.shape}")
        else:
            X_processed = self.preprocessor.transform(X)
        
        # Handle outliers only for numeric features
        try:
            X_numeric = X[self.numeric_features].copy()
            sample_weights = self.detect_outliers(X_numeric).squeeze()
            print("Outlier detection completed")
        except Exception as e:
            print(f"Error in outlier detection: {e}")
            sample_weights = np.ones(len(X))
        
        return X_processed, sample_weights
    
    def build_model(self, input_dim):
        """Build neural network model"""
        print(f"\nBuilding model with input dimension: {input_dim}")
        
        # Set random seed
        keras.utils.set_random_seed(self.random_state)
        
        # Input layers
        feature_input = layers.Input(shape=(input_dim,), name='features')
        weight_input = layers.Input(shape=(), name='sample_weights')
        
        # Main network
        x = layers.Dropout(0.2)(feature_input)
        
        # Hidden layers
        x = layers.Dense(64, activation='relu', 
                      kernel_regularizer=keras.regularizers.l2(0.01))(x)
        x = layers.BatchNormalization()(x)
        x = layers.Dropout(0.3)(x)
        
        x = layers.Dense(32, activation='relu',
                      kernel_regularizer=keras.regularizers.l2(0.01))(x)
        x = layers.BatchNormalization()(x)
        x = layers.Dropout(0.2)(x)
        
        x = layers.Dense(16, activation='relu',
                      kernel_regularizer=keras.regularizers.l2(0.01))(x)
        x = layers.BatchNormalization()(x)
        x = layers.Dropout(0.1)(x)
        
        # Output layer
        output = layers.Dense(1)(x)
        
        # Create model
        model = models.Model(inputs=[feature_input, weight_input], 
                           outputs=output)
        
        # Custom weighted loss function
        def weighted_mse(y_true, y_pred):
            sample_weights = weight_input
            return keras.backend.mean(
                sample_weights * keras.backend.square(y_true - y_pred)
            )
        
        # Compile model
        optimizer = keras.optimizers.Adam(learning_rate=0.001)
        model.compile(optimizer=optimizer,
                     loss=weighted_mse,
                     metrics=['mae'])
        
        self.model = model
        return model
    
    def fit(self, X, y, validation_split=0.2, epochs=100, batch_size=32, patience=10):
        """Fit the model"""
        print("\nStarting model fitting...")
        
        # Prepare data
        X_processed, sample_weights = self.prepare_data(X, y)
        
        # Convert y to numpy if needed
        y = y.values if hasattr(y, 'values') else np.array(y)
        
        # Build model if not exists
        if self.model is None:
            self.build_model(input_dim=X_processed.shape[1])
        
        # Create callbacks
        callbacks_list = [
            callbacks.EarlyStopping(
                monitor='val_loss',
                patience=patience,
                restore_best_weights=True
            ),
            callbacks.ReduceLROnPlateau(
                monitor='val_loss',
                factor=0.5,
                patience=patience//2,
                min_lr=0.00001
            )
        ]
        
        # Split data
        train_idx, val_idx = train_test_split(
            np.arange(len(X_processed)),
            test_size=validation_split,
            random_state=self.random_state
        )
        
        # Prepare training and validation data
        X_train = X_processed[train_idx]
        y_train = y[train_idx]
        w_train = sample_weights[train_idx]
        
        X_val = X_processed[val_idx]
        y_val = y[val_idx]
        w_val = sample_weights[val_idx]
        
        print("\nTraining shapes:")
        print(f"X_train: {X_train.shape}")
        print(f"y_train: {y_train.shape}")
        print(f"w_train: {w_train.shape}")
        
        # Fit model
        self.history = self.model.fit(
            [X_train, w_train], y_train,
            validation_data=([X_val, w_val], y_val),
            epochs=epochs,
            batch_size=batch_size,
            callbacks=callbacks_list,
            verbose=1
        )
        
        return self.history
    
    def predict(self, X):
        """Make predictions"""
        # Prepare data
        X_processed, sample_weights = self.prepare_data(X)
        
        # Make predictions
        predictions = self.model.predict([X_processed, sample_weights])
        
        return predictions
    
    def evaluate(self, X, y):
        """Evaluate model performance"""
        # Prepare data
        X_processed, sample_weights = self.prepare_data(X)
        
        # Convert y to numpy if needed
        y = y.values if hasattr(y, 'values') else np.array(y)
        
        # Evaluate
        scores = self.model.evaluate(
            [X_processed, sample_weights],
            y,
            verbose=0
        )
        
        metrics = dict(zip(self.model.metrics_names, scores))
        return metrics

# Example usage:
"""
model = RobustNeuralRegression()
history = model.fit(X, y)
predictions = model.predict(X_test)
metrics = model.evaluate(X_test, y_test)
"""
```

# Model comparison

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import (LinearRegression, HuberRegressor, 
                                TheilSenRegressor, RANSACRegressor)
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.svm import SVR
from sklearn.metrics import mean_squared_error, r2_score
import matplotlib.pyplot as plt
import seaborn as sns
from xgboost import XGBRegressor
from lightgbm import LGBMRegressor

class AlternativeModels:
    def __init__(self, random_state=42):
        self.random_state = random_state
        
    def _ensure_1d(self, y):
        """Ensure target variable is 1-dimensional"""
        if isinstance(y, pd.DataFrame):
            y = y.values
        if isinstance(y, pd.Series):
            y = y.values
        if len(y.shape) > 1:
            y = y.ravel()
        return y
        
    def compare_models(self, X, y, cv=5):
        """
        Compare different modeling approaches
        """
        # Ensure y is 1-dimensional
        y = self._ensure_1d(y)
        
        # Split data
        X_train, X_test, y_train, y_test = train_test_split(
            X, y, test_size=0.2, random_state=self.random_state
        )
        
        # Scale features
        scaler = StandardScaler()
        X_train_scaled = scaler.fit_transform(X_train)
        X_test_scaled = scaler.transform(X_test)
        
        # Define models to compare
        models = {
            'Linear Regression': LinearRegression(),
            
            # Robust Linear Models
            'Huber': HuberRegressor(epsilon=1.35),
            'RANSAC': RANSACRegressor(random_state=self.random_state),
            'TheilSen': TheilSenRegressor(random_state=self.random_state),
            
            # Tree-based Models
            'Random Forest': RandomForestRegressor(n_estimators=100, 
                                                 random_state=self.random_state),
            'Gradient Boosting': GradientBoostingRegressor(n_estimators=100,
                                                         random_state=self.random_state),
            'XGBoost': XGBRegressor(n_estimators=100, 
                                   random_state=self.random_state),
            'LightGBM': LGBMRegressor(n_estimators=100,
                                     random_state=self.random_state),
            
            # Support Vector Regression with different kernels
            'SVR (linear)': SVR(kernel='linear'),
            'SVR (rbf)': SVR(kernel='rbf'),
            'SVR (poly)': SVR(kernel='poly', degree=2)
        }
        
        results = {}
        feature_importance = {}
        predictions = {}
        
        for name, model in models.items():
            try:
                # Fit model
                model.fit(X_train_scaled, y_train)
                
                # Make predictions
                y_pred = model.predict(X_test_scaled)
                predictions[name] = y_pred
                
                # Calculate metrics
                r2 = r2_score(y_test, y_pred)
                rmse = np.sqrt(mean_squared_error(y_test, y_pred))
                mse = mean_squared_error(y_test, y_pred)
                
                # Perform cross-validation
                cv_scores = cross_val_score(model, X_train_scaled, y_train, 
                                          cv=cv, scoring='r2')
                
                results[name] = {
                    'R2': r2,
                    'MSE': mse,
                    'RMSE': rmse,
                    'CV_mean': cv_scores.mean(),
                    'CV_std': cv_scores.std()
                }
                
                # Get feature importance if available
                if hasattr(model, 'feature_importances_'):
                    feature_importance[name] = model.feature_importances_
                elif hasattr(model, 'coef_'):
                    if len(model.coef_.shape) > 1:
                        feature_importance[name] = np.abs(model.coef_[0])
                    else:
                        feature_importance[name] = np.abs(model.coef_)
                        
            except Exception as e:
                print(f"Error with {name}: {str(e)}")
                continue
        
        # Plot results
        self._plot_results(results)
        
        # Plot predictions
        self._plot_predictions(y_test, predictions)
        
        # Plot feature importance
        if feature_importance:
            self._plot_feature_importance(feature_importance, 
                                        X.columns if hasattr(X, 'columns') else None)
        
        return results, feature_importance
    
    def _plot_results(self, results):
        """Plot model comparison results"""
        if not results:
            print("No results to plot")
            return
            
        df_results = pd.DataFrame(results).T
        
        plt.figure(figsize=(15, 5))
        
        # R2 scores
        plt.subplot(131)
        df_results['R2'].plot(kind='bar')
        plt.title('R² Score by Model')
        plt.xlabel('Model')
        plt.ylabel('R²')
        plt.xticks(rotation=45)
        
        # RMSE
        plt.subplot(132)
        df_results['RMSE'].plot(kind='bar')
        plt.title('RMSE by Model')
        plt.xlabel('Model')
        plt.ylabel('RMSE')
        plt.xticks(rotation=45)
        
        # Cross-validation scores
        plt.subplot(133)
        plt.errorbar(range(len(df_results)), 
                    df_results['CV_mean'], 
                    yerr=df_results['CV_std'], 
                    fmt='o')
        plt.title('Cross-validation Scores')
        plt.xlabel('Model')
        plt.ylabel('CV Score')
        plt.xticks(range(len(df_results)), 
                  df_results.index, 
                  rotation=45)
        
        plt.tight_layout()
        plt.show()
    
    def _plot_predictions(self, y_test, predictions):
        """Plot predictions vs actual values"""
        if not predictions:
            print("No predictions to plot")
            return
            
        n_models = len(predictions)
        n_cols = 3
        n_rows = (n_models + n_cols - 1) // n_cols
        
        fig, axes = plt.subplots(n_rows, n_cols, 
                                figsize=(15, 5*n_rows))
        axes = axes.ravel()
        
        for i, (name, pred) in enumerate(predictions.items()):
            axes[i].scatter(y_test, pred, alpha=0.5)
            axes[i].plot([y_test.min(), y_test.max()], 
                        [y_test.min(), y_test.max()], 
                        'r--', lw=2)
            axes[i].set_title(name)
            axes[i].set_xlabel('Actual Values')
            axes[i].set_ylabel('Predicted Values')
        
        # Hide empty subplots
        for j in range(i + 1, len(axes)):
            axes[j].set_visible(False)
            
        plt.tight_layout()
        plt.show()
    
    def _plot_feature_importance(self, feature_importance, feature_names=None):
        """Plot feature importance comparison"""
        if not feature_importance:
            print("No feature importance to plot")
            return
            
        if feature_names is None:
            feature_names = [f'Feature_{i}' for i in range(len(next(iter(feature_importance.values()))))]
            
        importance_df = pd.DataFrame(feature_importance, index=feature_names)
        
        # Normalize importance scores
        for column in importance_df.columns:
            importance_df[column] = importance_df[column] / importance_df[column].sum()
        
        plt.figure(figsize=(12, 6))
        sns.heatmap(importance_df, annot=True, cmap='YlOrRd')
        plt.title('Normalized Feature Importance by Model')
        plt.xlabel('Model')
        plt.ylabel('Feature')
        plt.xticks(rotation=45)
        plt.tight_layout()
        plt.show()
        
        return importance_df
```
